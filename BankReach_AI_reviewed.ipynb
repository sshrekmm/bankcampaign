```python
budgets = [500, 1000, 2000, 5000]

simulation_results = []

for budget in budgets:

    selected = customer_priority.head(budget)

    actual_subscribers = selected['Actual'].sum()

    response_rate = (
        actual_subscribers / budget
    ) * 100

    total_subscribers = customer_priority['Actual'].sum()

    subscriber_capture = (
        actual_subscribers / total_subscribers
    ) * 100

    simulation_results.append({
        'Calling_Budget': budget,
        'Actual_Subscribers': actual_subscribers,
        'Response_Rate': round(response_rate, 2),
        'Subscriber_Capture': round(subscriber_capture, 2)
    })

simulation_df = pd.DataFrame(simulation_results)

print(simulation_df)
```

```python
plt.figure(figsize=(9, 5))

sns.lineplot(
    data=simulation_df,
    x='Calling_Budget',
    y='Response_Rate',
    marker='o'
)

plt.title("Calling Budget vs Response Rate")
plt.xlabel("Number of Customers Contacted")
plt.ylabel("Response Rate (%)")
plt.grid()
plt.show()
```

```python
budgets = [500, 1000, 2000, 5000]

comparison_results = []

total_subscribers = customer_results['Actual'].sum()

for budget in budgets:

    # AI-based selection
    ai_selected = customer_priority.head(budget)

    ai_subscribers = ai_selected['Actual'].sum()

    ai_response_rate = (ai_subscribers / budget) * 100

    ai_capture = (ai_subscribers / total_subscribers) * 100

    # Random selection: repeat 100 times
    random_subscribers = []

    for i in range(100):
        random_selected = customer_results.sample(
            n=budget,
            random_state=42 + i
        )

        random_subscribers.append(
            random_selected['Actual'].sum()
        )

    avg_random_subscribers = np.mean(random_subscribers)

    random_response_rate = (
        avg_random_subscribers / budget
    ) * 100

    random_capture = (
        avg_random_subscribers / total_subscribers
    ) * 100

    comparison_results.append({
        'Calling_Budget': budget,
        'AI_Subscribers': ai_subscribers,
        'AI_Response_Rate': round(ai_response_rate, 2),
        'AI_Capture': round(ai_capture, 2),
        'Random_Subscribers_Avg': round(avg_random_subscribers, 2),
        'Random_Response_Rate': round(random_response_rate, 2),
        'Random_Capture': round(random_capture, 2)
    })

comparison_df = pd.DataFrame(comparison_results)

print(comparison_df)
```

```python
plt.figure(figsize=(9, 5))

plt.plot(
    comparison_df['Calling_Budget'],
    comparison_df['AI_Response_Rate'],
    marker='o',
    label='AI Prioritization'
)

plt.plot(
    comparison_df['Calling_Budget'],
    comparison_df['Random_Response_Rate'],
    marker='o',
    label='Random Selection'
)

plt.title("AI vs Random Customer Selection")
plt.xlabel("Number of Customers Contacted")
plt.ylabel("Response Rate (%)")

plt.legend()
plt.grid()
plt.show()
```

```python
plt.figure(figsize=(9, 5))

plt.plot(
    comparison_df['Calling_Budget'],
    comparison_df['AI_Capture'],
    marker='o',
    label='AI Prioritization'
)

plt.plot(
    comparison_df['Calling_Budget'],
    comparison_df['Random_Capture'],
    marker='o',
    label='Random Selection'
)

plt.title("AI vs Random Subscriber Capture")
plt.xlabel("Number of Customers Contacted")
plt.ylabel("Subscribers Captured (%)")

plt.legend()
plt.grid()
plt.show()
```

```python
comparison_df['Response_Rate_Improvement'] = (
    (comparison_df['AI_Response_Rate'] -
     comparison_df['Random_Response_Rate'])
    / comparison_df['Random_Response_Rate']
) * 100

comparison_df['Additional_Subscribers'] = (
    comparison_df['AI_Subscribers'] -
    comparison_df['Random_Subscribers_Avg']
)

print(comparison_df[[
    'Calling_Budget',
    'AI_Response_Rate',
    'Random_Response_Rate',
    'Response_Rate_Improvement',
    'Additional_Subscribers'
]])
```

```python
def score_new_customers(new_customers, model_bundle):
    # Work on a copy so the original input is unchanged.
    new_data = new_customers.copy()

    # Remove fields that must not be used for pre-call prediction.
    new_data = new_data.drop(columns=['y', 'duration'], errors='ignore')

    # Encode categorical columns in the new records.
    new_encoded = pd.get_dummies(new_data, dtype=int)

    # Add any training columns missing from the new data and keep the same order.
    feature_columns = model_bundle['feature_columns']
    new_encoded = new_encoded.reindex(columns=feature_columns, fill_value=0)

    # Estimate probability of subscription (class 1 = yes).
    probabilities = model_bundle['model'].predict_proba(new_encoded)[:, 1]

    scored = new_customers.copy()
    scored['Subscription_Probability'] = probabilities
    scored['Subscription_Probability_Percent'] = (probabilities * 100).round(2)
    scored['Priority_Rank'] = (
        scored['Subscription_Probability']
        .rank(method='first', ascending=False)
        .astype(int)
    )

    return scored.sort_values('Subscription_Probability', ascending=False)
```
